Q11. HR wants to know whether age and monthly income are related. Create a visualization
showing their relationship and explain whether income generally increases with age.

```python
m , b = np.polyfit(df['Age'],df['MonthlyIncome'],1)
plt.scatter(df['Age'],df['MonthlyIncome'],color='#36415A',s=60,edgecolors='#202734')


plt.plot(df['Age'],
         m * df['Age']+b,
         color='red',
         label = 'Trend Line',
         linestyle=':'
         )


plt.title("Relation between age and monthly income")
plt.xlabel("Age")
plt.ylabel("Monthly Income")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

The upward trend line suggests that monthly income generally increases with age. However, incomes vary widely among employees of the same age, so age alone does not explain salary differences.

Q12. Management wants to understand whether employees with different job roles receive

Q13. HR suspects overtime may affect employees differently by department. Create a
visualization comparing attrition across departments, separated by overtime status. Which
employee groups appear most vulnerable?

```python
df.columns
```

```python


data = df.copy()
data["Left"] = data["Attrition"] == "Yes"

rates = data.groupby(["Department", "OverTime"])["Left"].mean().unstack() *100      ##True counts as 1 and False as 0,



ax = rates.plot(kind="bar", figsize=(9, 5),color=['#B8892D','#D8C9A8'])

ax.set_title("Attrition Rate by Department and Overtime")
ax.set_xlabel("Department")
ax.set_ylabel("Attrition rate (%)")
ax.set_ylim(0, rates.max().max() * 1.2)
ax.legend(title="Works overtime")
ax.tick_params(axis="x", rotation=0)

# Show the percentage above each bar
for bars in ax.containers:
    ax.bar_label(bars, fmt="%.1f%%")

plt.tight_layout()
plt.show()
```

Q17. The leadership team suspects that employees with longer tenure and higher salaries
may be less likely to leave, but overtime could change this relationship. Create a
visualization incorporating years at the company, monthly income, overtime, and attrition.
Explain the patterns you observe.

```python


fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharex=True, sharey=True)

for x, overtime in zip(axes, ["No", "Yes"]):
    group = df[df["OverTime"] == overtime]              ##zip() pairs items from two sequences by position. Here it pairs:

    stayed = df[df["Attrition"] == "No"]
    left = df[df["Attrition"] == "Yes"]

    x.scatter(stayed["YearsAtCompany"], stayed["MonthlyIncome"],
               color="blue", label="Stayed", alpha=0.5)
    x.scatter(left["YearsAtCompany"], left["MonthlyIncome"],
               color="red", label="Left", alpha=0.5)

    x.set_title(f"Overtime: {overtime}")
    x.set_xlabel("Years at company")

axes[0].set_ylabel("Monthly income")
axes[0].legend()
plt.tight_layout()
plt.show()
```

Employees who left were often newer and earned less, especially if they worked overtime. Employees who stayed were more common among those with longer tenure and higher incomes.

* Main patterns: Attrition appears higher among employees who work overtime and among employees with lower job satisfaction or poorer work-life balance. Employees who leave also appear more concentrated in lower-income groups. The age–income plot shows income generally rising with age, but employees of the same age have widely varying incomes.

* Groups to investigate: Employees who work overtime, employees with low satisfaction or poor work-life balance, and lower-paid employees. 

* Evidence: Use the grouped bar chart to compare overtime attrition rates, the box plot and grouped bars to identify satisfaction and work-life balance combinations, and the scatter plots to show how age, income, tenure, and attrition relate. 

* Recommended actions: Review overtime workloads; ask employees about job satisfaction and work-life balance; and examine pay, career development, and support in groups with high attrition. Monitor attrition rates after making changes.

* Limitations: These charts show associations, not causes. Small groups can have unstable rates, and job role, manager, or other factors may affect both pay and attrition. Statistical tests would be needed to confirm whether group differences are significant.
